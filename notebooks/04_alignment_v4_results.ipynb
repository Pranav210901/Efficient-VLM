# Alignment v4 — Distillation Capture Probe

**Main dissertation question:** Can modern frozen SSL vision encoders, combined with compatibility-based pair selection and sub-5M-parameter adaptation, approach compact jointly pretrained vision-language models under a fixed inference budget?

This notebook reports the v4 diagnostic probe. A viable probe selects the next methodology; it does not by itself establish the dissertation claim.

```python
import json
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().resolve()
if not (ROOT / 'results').is_dir() and (ROOT.parent / 'results').is_dir():
    ROOT = ROOT.parent
REPORT = ROOT / 'results/alignment_v4/combined_probe_report.json'
payload = json.loads(REPORT.read_text())
print(payload['main_dissertation_question'])
print('Overall probe decision:', payload['overall_probe_decision'])
print(payload['claim_status'])
```

```python
frame = pd.DataFrame(payload['teachers'])
display(frame[['teacher_label', 'baseline_R@1', 'distilled_R@1', 'teacher_R@1', 'delta_R@1', 'capture_fraction', 'decision']])
```

```python
ax = frame.set_index('teacher_label')[['baseline_R@1', 'distilled_R@1', 'teacher_R@1']].plot.bar(figsize=(9, 4))
ax.axhline(0.30, color='black', linestyle='--', linewidth=1, label='viability threshold')
ax.set_ylabel('Bidirectional mean R@1')
ax.set_title('Matched baseline, distilled student, and teacher anchor')
ax.legend()
plt.tight_layout()
plt.show()
```

## Interpretation guardrails

- `distilled_R@1 ≥ 0.30`: the frozen-encoder absolute-performance route remains viable.
- `distilled_R@1 ≤ 0.22`: pivot the dissertation emphasis to efficiency-normalized performance.
- Between these values: the probe is inconclusive.
- Capture fraction is directional because teacher anchors and student probes use different disjoint splits; the absolute student threshold is the primary gate.
